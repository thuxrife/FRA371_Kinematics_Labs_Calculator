```python
import spatialgeometry as sg
import numpy as np
import spatialmath as sm
from Lab3_env_check_utils import *
```

```python
urdf_file_name = 'robot.urdf'  # Change this to the desired URDF file
robot = DefindeRobot(urdf_file_name=urdf_file_name)  

total_time = 5.0   # [s]
current_time = 0.0  # [s]
dt = 0.01 # [s]

env2 = SetupSimulation(robot=robot, start_position=[0.5, -0.4, 0.5])

box = sg.Cuboid(
    scale=[0.1, 0.1, 0.1], 
    pose=sm.SE3(0.6, 0.3, 0.4),
    color=[0.65, 0.65, 0.65, 1.0]     # RGBA color format (Gray, solid)
)
env2.add(box)

M_hist = []
q_dot_hist = []
err_hist = []
timestamp = []

while current_time < total_time:
    current_box_position = sm.SE3(box.T).t
    
    robot.qd = np.array([0.0, 0.0, 0.1])
    V=np.array([-0.05, 0.0, 0.0])
    UpdateBoxPosition(box=box, P_current=current_box_position, V=V, dt=dt) 

    env2.step(dt)
    timestamp.append(current_time)
    current_time += dt
```
